### **Part 1. Imports and data loading**




Importing packages and libraries needed to compute descriptive statistics and conduct a hypothesis test.

```python
# Loading dataset into dataframe
df = pd.read_csv('waze_dataset.csv')
```

## **PACE: Analyze and Construct**

### **Part 2. Data exploration**

Using descriptive statistics to conduct exploratory data analysis (EDA).

In the dataset, `device` is a categorical variable with the labels `iPhone` and `Android`.

In order to perform this analysis, you must turn each label into an integer.  The following code assigns a `1` for an `iPhone` user and a `2` for `Android`.  It assigns this label back to the variable `device_new`, creating a new variable so that we don't overwrite original data.



```python
# 1. Creating `map_dictionary`
map_dictionary = {'iPhone':1,'Android':2}
# 2. Creating new `device_type` column
df['device_type'] = df['device'].map(map_dictionary)
# 3. Mapping the new column to the dictionary
df.head()
```

We are interested in the relationship between device type and the number of drives. One approach is to look at the average number of drives for each device type. Calculating these averages.

```python
df.groupby('device')['drives'].mean()
```

Based on the averages shown, it appears that drivers who use an iPhone device to interact with the application have a higher number of drives on average. However, this difference might arise from random sampling, rather than being a true difference in the number of drives. To assess whether the difference is statistically significant, we can conduct a hypothesis test.


### **Part 3. Hypothesis testing**

Our goal is to conduct a two-sample t-test. The steps for conducting a hypothesis test:


1.   Statinh the null hypothesis and the alternative hypothesis
2.   Choosing a signficance level
3.   Finding the p-value
4.   Rejecting or fail to reject the null hypothesis

This is a t-test for two independent samples. This is the appropriate test since the two groups are independent (Android users vs. iPhone users).

**Null Hypothesis:** There is no effect of the user's device type on number of rides . <br>
**Alternative Hypothesis:** There is an effect of the user's device type on number of rides .

Next, we choosed 5% as the significance level and proceed with a two-sample t-test.

We can use the `stats.ttest_ind()` function to perform the test.


The default for the argument `equal_var` in `stats.ttest_ind()` is `True`, which assumes population variances are equal. This equal variance assumption might not hold in practice (that is, there is no strong reason to assume that the two groups have the same variance); we can relax this assumption by setting `equal_var` to `False`, and `stats.ttest_ind()` will perform the unequal variances $t$-test (known as Welch's `t`-test).


1. Isolating the `drives` column for iPhone users.
2. Isolating the `drives` column for Android users.
3. Performing the t-test

```python
iphone = df[df['device_type'] ==1]['drives']
android= df[df['device_type']==2]['drives']
stats.ttest_ind(a=iphone,b=android,equal_var=False)
```

We can see thap `p-value = 14.33% ` which is greater than the significance level of 5%. Therefore, we `fail` to regect the null hypothesis.
